<h2 id="setup" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">⚙️ 1. Setup &amp; Dashboard Styling</h2>

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.patches import FancyBboxPatch
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# ---- "Social Pulse" light glassmorphic palette ----
BG        = '#F7F9FC'
PANEL     = '#FFFFFF'
INK       = '#1E293B'
INK_SOFT  = '#64748B'
TWITTER   = '#1DA1F2'
YOUTUBE   = '#FF0000'
REDDIT    = '#FF4500'
GREEN     = '#22C55E'
AMBER     = '#F59E0B'
RED       = '#EF4444'
PURPLE    = '#8B5CF6'
GRID      = '#E1E7F0'

PLATFORM_COLORS = {'Twitter': TWITTER, 'YouTube': YOUTUBE, 'Reddit': REDDIT}
SENTIMENT_COLORS = {'Positive': GREEN, 'Neutral': '#94A3B8', 'Negative': RED}
PALETTE_CAT = [TWITTER, REDDIT, AMBER, GREEN, PURPLE, RED, '#0EA5E9']

plt.rcParams.update({
    'figure.facecolor': BG,
    'axes.facecolor': PANEL,
    'savefig.facecolor': BG,
    'axes.edgecolor': GRID,
    'axes.labelcolor': INK,
    'text.color': INK,
    'xtick.color': INK_SOFT,
    'ytick.color': INK_SOFT,
    'axes.grid': True,
    'grid.color': GRID,
    'grid.linewidth': 0.8,
    'grid.alpha': 0.8,
    'font.family': 'sans-serif',
    'font.size': 11,
    'axes.titlesize': 14,
    'axes.titleweight': 'bold',
    'axes.spines.top': False,
    'axes.spines.right': False,
})

sns.set_palette(sns.color_palette(PALETTE_CAT))

def style_ax(ax, title=None, xlabel=None, ylabel=None):
    if title: ax.set_title(title, color=INK, pad=12)
    if xlabel: ax.set_xlabel(xlabel, color=INK_SOFT)
    if ylabel: ax.set_ylabel(ylabel, color=INK_SOFT)
    ax.spines['left'].set_color(GRID)
    ax.spines['bottom'].set_color(GRID)
    return ax

print("Dashboard styling loaded.")
```

<h2 id="firstlook" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">📰 2. First Look at the Feed</h2>

```python

df = pd.read_csv('/kaggle/input/datasets/dreamtensor/social-media-sentiment-analysisai-trend-2026/Social_Media_Sentiment_Analysis_AI_Trends_2026.csv')
df['posted_datetime'] = pd.to_datetime(df['posted_datetime'])
print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns\n")
df.head(6)
```

<h2 id="quality" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">🔍 3. Data Quality &amp; a Reality Check</h2>
<p style="color:#64748B; font-family:'Segoe UI',Helvetica,Arial,sans-serif;">Before charting anything, it's worth checking whether the text actually drives the labels — because if it doesn't, every "sentiment by X" chart later needs a caveat.</p>

```python

print("Duplicate rows:", df.duplicated().sum())
print("\nNulls:")
print(df.isnull().sum()[df.isnull().sum() > 0], "\n(mentions is null whenever a post doesn't @-mention anyone — expected, not a defect.)")

print(f"\nUnique post_text values: {df['post_text'].nunique()} (out of {len(df):,} rows)")
```

```python

# Does the actual text content predict the assigned sentiment label?
text_sentiment = pd.crosstab(df['post_text'], df['sentiment_label'], normalize='index').round(2)
print("Share of each sentiment label, per template sentence:")
text_sentiment
```

```python

fig, ax = plt.subplots(figsize=(11, 5.5))
text_sentiment[['Negative','Neutral','Positive']].plot(
    kind='barh', stacked=True, ax=ax,
    color=[SENTIMENT_COLORS['Negative'], SENTIMENT_COLORS['Neutral'], SENTIMENT_COLORS['Positive']],
    edgecolor=BG, linewidth=1.2)
ax.axvline(1/3, color=INK, linestyle=':', linewidth=1)
style_ax(ax, 'Sentiment Label Share by Post Text — Nearly Identical for Every Sentence', 'Share of posts', '')
ax.legend(title='', frameon=False, fontsize=9, loc='lower right')
plt.tight_layout()
plt.show()
```


<div style="background:#FFF7ED; border-left:3px solid #F59E0B; padding:14px 20px; font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#78350F; font-size:14px; border-radius:6px;">
<b>Important finding, stated plainly:</b> the dataset draws from just <b>10 template sentences</b> repeated across
all 2,200 posts (e.g. <i>"Worst decision ever made."</i> and <i>"This is absolutely amazing!"</i> each appear ~200+
times). Yet every template gets Negative/Neutral/Positive labels in almost exactly a 1/3–1/3–1/3 split, regardless
of what the sentence actually says — a clearly negative line like "Worst decision ever made" is labeled Positive
about as often as Negative. <b>Sentiment labels here are assigned independently of the text</b>, which strongly
suggests this is a synthetic dataset for practicing analysis pipelines, not real scraped social posts. We'll treat
every "sentiment vs. X" chart below as a pattern in the <i>labels</i>, not a reflection of real opinion.
</div>


<h2 id="volume" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">📊 4. Volume: Platforms, Time &amp; Language</h2>

```python

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

ax = axes[0]
plat_counts = df['platform'].value_counts()
bars = ax.bar(plat_counts.index, plat_counts.values,
              color=[PLATFORM_COLORS[p] for p in plat_counts.index], edgecolor=BG, linewidth=1.5, width=0.6)
for b in bars:
    ax.text(b.get_x()+b.get_width()/2, b.get_height()+8, int(b.get_height()), ha='center', fontsize=10, color=INK)
style_ax(ax, 'Posts by Platform', '', 'Count')

ax = axes[1]
monthly = df.set_index('posted_datetime').resample('MS').size()
ax.plot(monthly.index, monthly.values, color=TWITTER, linewidth=2, marker='o', markersize=3)
style_ax(ax, 'Posts per Month', 'Month', 'Count')

ax = axes[2]
lang_counts = df['language'].value_counts()
bars = ax.barh(lang_counts.index[::-1], lang_counts.values[::-1], color=PALETTE_CAT[:len(lang_counts)][::-1], edgecolor=BG, linewidth=1.2)
style_ax(ax, 'Posts by Language', 'Count', '')

plt.tight_layout()
plt.show()

print("Post volume is close to evenly split across platforms, months, and languages — another sign of a")
print("deliberately balanced, simulated sample rather than organically scraped social data.")
```

<h2 id="sentiment" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">💬 5. The Sentiment Landscape</h2>

```python

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

ax = axes[0]
sent_counts = df['sentiment_label'].value_counts().reindex(['Positive','Neutral','Negative'])
bars = ax.bar(sent_counts.index, sent_counts.values,
              color=[SENTIMENT_COLORS[s] for s in sent_counts.index], edgecolor=BG, linewidth=1.5, width=0.55)
for b in bars:
    ax.text(b.get_x()+b.get_width()/2, b.get_height()+8, int(b.get_height()), ha='center', fontsize=10, color=INK)
style_ax(ax, 'Overall Sentiment Split', '', 'Count')

ax = axes[1]
ax.hist(df['sentiment_score'], bins=40, color=TWITTER, edgecolor=BG, linewidth=0.5)
ax.axvline(-0.3, color=RED, linestyle='--', linewidth=1)
ax.axvline(0.3, color=GREEN, linestyle='--', linewidth=1)
style_ax(ax, 'Sentiment Score Distribution', 'Score (-1 to +1)', 'Count')

ax = axes[2]
plat_sent = pd.crosstab(df['platform'], df['sentiment_label'], normalize='index')[['Positive','Neutral','Negative']]
plat_sent.plot(kind='bar', stacked=True, ax=ax,
               color=[SENTIMENT_COLORS['Positive'], SENTIMENT_COLORS['Neutral'], SENTIMENT_COLORS['Negative']],
               edgecolor=BG, linewidth=1.2)
style_ax(ax, 'Sentiment Share by Platform', '', 'Share')
ax.legend(title='', frameon=False, fontsize=8)
plt.setp(ax.get_xticklabels(), rotation=0)

plt.tight_layout()
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(11, 5.5))
topic_sent = pd.crosstab(df['topic_category'], df['sentiment_label'], normalize='index')[['Positive','Neutral','Negative']]
topic_sent = topic_sent.loc[topic_sent['Positive'].sort_values(ascending=False).index]
topic_sent.plot(kind='barh', stacked=True, ax=ax,
                 color=[SENTIMENT_COLORS['Positive'], SENTIMENT_COLORS['Neutral'], SENTIMENT_COLORS['Negative']],
                 edgecolor=BG, linewidth=1.2)
style_ax(ax, 'Sentiment Share by Topic Category', 'Share', '')
ax.legend(title='', frameon=False, fontsize=9, loc='lower right')
plt.tight_layout()
plt.show()

print("Every platform and topic hovers within a few points of the ~33/33/33 baseline — consistent with the")
print("earlier finding that sentiment labels aren't tied to real content.")
```

<h2 id="emotion" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">😐 6. Emotion, Toxicity &amp; Sarcasm</h2>

```python

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

ax = axes[0]
emo_counts = df['emotion_label'].value_counts()
emo_colors = {'Happy': GREEN, 'Neutral': '#94A3B8', 'Sad': TWITTER, 'Fear': PURPLE, 'Angry': RED}
bars = ax.bar(emo_counts.index, emo_counts.values, color=[emo_colors.get(e,'#999') for e in emo_counts.index], edgecolor=BG, linewidth=1.5)
for b in bars:
    ax.text(b.get_x()+b.get_width()/2, b.get_height()+6, int(b.get_height()), ha='center', fontsize=9, color=INK)
style_ax(ax, 'Emotion Label Distribution', '', 'Count')

ax = axes[1]
ax.hist(df['toxicity_score'], bins=30, color=RED, edgecolor=BG, linewidth=0.5, alpha=0.85)
style_ax(ax, 'Toxicity Score Distribution', 'Toxicity (0-1)', 'Count')

ax = axes[2]
flags = pd.DataFrame({
    'Sarcasm': df['sarcasm_detected'].value_counts(normalize=True)*100,
    'Spam': df['spam_flag'].value_counts(normalize=True)*100
}).T
flags = flags[[True, False]]
flags.columns = ['Flagged', 'Not flagged']
flags.plot(kind='barh', stacked=True, ax=ax, color=[AMBER, '#E2E8F0'], edgecolor=BG, linewidth=1.2)
style_ax(ax, 'Sarcasm &amp; Spam Flag Rates', '%', '')
ax.legend(frameon=False, fontsize=9)

plt.tight_layout()
plt.show()

print(f"Toxicity score is close to uniformly distributed (mean={df['toxicity_score'].mean():.2f}), and roughly")
print(f"{df['sarcasm_detected'].mean()*100:.0f}% of posts are flagged for sarcasm, {df['spam_flag'].mean()*100:.0f}% for spam —")
print("both close to a coin flip, again pointing to randomly assigned rather than model-derived flags.")
```

<h2 id="engagement" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">📈 7. Engagement Metrics</h2>

```python

fig, axes = plt.subplots(2, 2, figsize=(15, 10.5))

ax = axes[0,0]
ax.hist(df['like_count'], bins=30, color=TWITTER, edgecolor=BG, linewidth=0.5)
style_ax(ax, 'Like Count Distribution', 'Likes', 'Count of posts')

ax = axes[0,1]
ax.hist(df['engagement_score'], bins=30, color=GREEN, edgecolor=BG, linewidth=0.5)
style_ax(ax, 'Engagement Score Distribution', 'Score', 'Count of posts')

ax = axes[1,0]
verified_eng = df.groupby('user_verified')[['like_count','comment_count','share_count']].mean()
verified_eng.index = ['Not verified', 'Verified']
verified_eng.plot(kind='bar', ax=ax, color=[TWITTER, AMBER, REDDIT], edgecolor=BG, linewidth=1.2)
style_ax(ax, 'Avg. Engagement: Verified vs. Not', '', 'Average count')
plt.setp(ax.get_xticklabels(), rotation=0)
ax.legend(frameon=False, fontsize=9)

ax = axes[1,1]
trend_eng = df.groupby('is_trending_topic')[['like_count','comment_count','share_count']].mean()
trend_eng.index = ['Not trending', 'Trending']
trend_eng.plot(kind='bar', ax=ax, color=[TWITTER, AMBER, REDDIT], edgecolor=BG, linewidth=1.2)
style_ax(ax, 'Avg. Engagement: Trending vs. Not', '', 'Average count')
plt.setp(ax.get_xticklabels(), rotation=0)
ax.legend(frameon=False, fontsize=9)

plt.tight_layout()
plt.show()

print("Verified accounts and trending-flagged posts show essentially the same average engagement as everyone")
print("else — in a real platform dataset this would be a notable (and unusual) finding; here it's consistent")
print("with the broader pattern of randomly assigned fields.")
```

<h2 id="corr" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">🔗 8. Correlation Check</h2>

```python

num_cols = ['user_followers_count','post_length','like_count','comment_count','share_count',
            'engagement_score','sentiment_score','toxicity_score']
corr = df[num_cols].corr()

fig, ax = plt.subplots(figsize=(8.5, 7))
cmap = sns.diverging_palette(10, 220, s=60, l=55, sep=20, as_cmap=True)
mask = np.triu(np.ones_like(corr, dtype=bool), k=1)
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap=cmap, center=0, vmin=-0.2, vmax=0.2,
            square=True, linewidths=1.2, linecolor=BG, cbar_kws={'shrink':0.8,'label':'Pearson r'},
            annot_kws={'fontsize':9}, ax=ax)
ax.set_title('Correlation Matrix — Numeric Fields', color=INK, pad=14)
plt.xticks(rotation=40, ha='right', fontsize=9)
plt.yticks(rotation=0, fontsize=9)
plt.tight_layout()
plt.show()

max_corr = corr.where(~np.eye(len(corr), dtype=bool)).abs().max().max()
print(f"Largest off-diagonal correlation in the entire matrix: {max_corr:.3f} — essentially noise.")
print("Even engagement_score barely relates to the raw like/comment/share counts it's presumably built from,")
print("reinforcing that these fields were generated independently rather than derived from one another.")
```

<h2 id="text" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">🏷️ 9. Text &amp; Hashtag Patterns</h2>

```python

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

ax = axes[0]
text_counts = df['post_text'].value_counts()
ax.barh([t[:32]+('…' if len(t)>32 else '') for t in text_counts.index[::-1]], text_counts.values[::-1],
        color=TWITTER, edgecolor=BG, linewidth=1.0)
style_ax(ax, 'The 10 Template Sentences, by Frequency', 'Count', '')

ax = axes[1]
hashtag_counts = df['hashtags'].value_counts()
ax.barh(hashtag_counts.index[::-1], hashtag_counts.values[::-1], color=REDDIT, edgecolor=BG, linewidth=1.2)
style_ax(ax, 'Hashtag Combinations Used', 'Count', '')

plt.tight_layout()
plt.show()

print(f"Only {df['hashtags'].nunique()} distinct hashtag combinations appear across all {len(df):,} posts —")
print("further evidence this is a templated dataset rather than free-text scraped content.")
```

<h2 id="dashboard" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">📋 10. Summary Dashboard</h2>

<h2 id="model" style="font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#1E293B; border-bottom:3px solid #1DA1F2; padding-bottom:8px;">🤖 11. Can We Predict Sentiment?</h2>
<div style="background:#FFF7ED; border-left:3px solid #F59E0B; padding:14px 20px; font-family:'Segoe UI',Helvetica,Arial,sans-serif; color:#78350F; font-size:14px; border-radius:6px; margin-bottom:14px;">
<b>Expectation, set honestly upfront:</b> given Section 3's finding that sentiment labels don't track the text or any
other field, a well-behaved classifier here should perform <b>close to the ~33% random-guess baseline</b> for three
balanced classes. If a model claimed 90% accuracy on this data, that would be a red flag for leakage, not a
success story — we're building this classifier to demonstrate the workflow and confirm the "no real signal" finding
quantitatively, not to ship a sentiment engine.
</div>

```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

model_df = df.copy()
cat_cols = ['platform', 'user_verified', 'user_location', 'language', 'hashtags',
            'day_of_week', 'is_trending_topic', 'topic_category', 'emotion_label',
            'sarcasm_detected', 'spam_flag']
for c in cat_cols:
    model_df[c] = LabelEncoder().fit_transform(model_df[c].astype(str))

features = ['platform', 'user_verified', 'user_followers_count', 'user_location', 'language',
            'hashtags', 'post_length', 'like_count', 'comment_count', 'share_count',
            'engagement_score', 'day_of_week', 'is_trending_topic', 'topic_category',
            'emotion_label', 'toxicity_score', 'sarcasm_detected', 'spam_flag']
X = model_df[features]
y = model_df['sentiment_label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)

clf = RandomForestClassifier(n_estimators=300, max_depth=8, random_state=42, n_jobs=-1)
clf.fit(X_train, y_train)
pred = clf.predict(X_test)
acc = accuracy_score(y_test, pred)

print(f"Random Forest accuracy: {acc:.1%}  (random-guess baseline for 3 balanced classes ≈ 33.3%)")
print()
print(classification_report(y_test, pred))
```

```python

cm = confusion_matrix(y_test, pred, labels=['Negative','Neutral','Positive'])
fig, ax = plt.subplots(figsize=(6.5, 5.5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', square=True, cbar=False,
            xticklabels=['Negative','Neutral','Positive'], yticklabels=['Negative','Neutral','Positive'],
            linewidths=1.5, linecolor=BG, annot_kws={'fontsize':12}, ax=ax)
style_ax(ax, f'Confusion Matrix — Random Forest (accuracy={acc:.1%})', 'Predicted', 'Actual')
plt.tight_layout()
plt.show()
```

<b>Confirmed:</b> the model lands right around the ~33% random-guess baseline, and the confusion matrix shows
