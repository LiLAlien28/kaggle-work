## 1. Setup & Data Loading <a id='1'></a>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.gridspec as gridspec
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.lines import Line2D
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Twilight / clinical dark theme ────────────────────────────────────────────
BG, PANEL, GRID = '#16161d', '#1a1a24', '#2e2e3a'

plt.rcParams.update({
    'figure.facecolor': BG,   'axes.facecolor': PANEL,
    'axes.edgecolor':   GRID, 'axes.labelcolor': '#9ca3b0',
    'axes.titlecolor':  '#f5f5fa', 'axes.titlesize': 13,
    'axes.labelsize':   11,   'xtick.color': '#5a5a70',
    'ytick.color':      '#5a5a70', 'text.color': '#9ca3b0',
    'grid.color': '#22222e', 'grid.linewidth': 0.6,
    'legend.facecolor': '#16161d', 'legend.edgecolor': GRID,
    'font.family': 'DejaVu Sans',
})

# Palette
DEP   = '#f87171'   # depression (PHQ-9)
ANX   = '#fb923c'   # anxiety (GAD-7)
STR   = '#fbbf24'   # stress
SLP   = '#c084fc'   # sleep
SCR   = '#38bdf8'   # screen time
ACT   = '#34d399'   # physical activity
ACA   = '#60a5fa'   # academic
FIN   = '#e879f9'   # financial

SEV_PAL = {
    'None':         '#3d4a5c',
    'Mild':         '#fbbf24',
    'Moderate':     '#fb923c',
    'Mod-Severe':   '#f87171',
    'Severe':       '#dc2626',
}
GENDER_PAL = {'Male': '#38bdf8', 'Female': '#f472b6', 'Other': '#a78bfa'}
MAJOR_PAL  = {
    'STEM': '#38bdf8', 'Business': '#fbbf24', 'Social Sciences': '#a78bfa',
    'Health Sciences': '#34d399', 'Humanities': '#f472b6',
    'Arts': '#fb923c', 'Law': '#60a5fa',
}
CMAP_DEP  = LinearSegmentedColormap.from_list('dep',  [PANEL, STR, DEP])
CMAP_COOL = LinearSegmentedColormap.from_list('cool', [PANEL, ACA, SLP])
CMAP_DIV  = LinearSegmentedColormap.from_list('div',  [ACT, PANEL, DEP])

print('✅  Twilight clinical theme active. Mental health data ready.')
```

## 3. Feature Engineering <a id='3'></a>

```python
# PHQ-9 severity categories
def phq9_severity(score):
    if pd.isna(score): return np.nan
    if score <= 4:  return 'None'
    if score <= 9:  return 'Mild'
    if score <= 14: return 'Moderate'
    if score <= 19: return 'Mod-Severe'
    return 'Severe'

def gad7_severity(score):
    if pd.isna(score): return np.nan
    if score <= 4:  return 'Minimal'
    if score <= 9:  return 'Mild'
    if score <= 14: return 'Moderate'
    return 'Severe'

df['phq9_severity'] = df['phq9_score'].apply(phq9_severity)
df['gad7_severity'] = df['gad7_score'].apply(gad7_severity)

# Any clinically significant distress flag
df['at_risk'] = ((df['phq9_score'] >= 10) | (df['gad7_score'] >= 10)).astype(int)

# Sleep adequacy
df['sleep_adequate'] = (df['daily_sleep_hours'] >= 7).astype(int)

# CGPA band
df['cgpa_band'] = pd.cut(df['cgpa'], bins=[0,2.5,3.0,3.5,4.5],
                          labels=['<2.5','2.5-3.0','3.0-3.5','3.5+'])

# Composite wellbeing index (higher = better)
df['wellbeing_idx'] = (
    (10 - df['phq9_score'].clip(0,10)) * 0.3
    + (7  - df['gad7_score'].clip(0,7)) * 0.3
    + df['daily_sleep_hours'].clip(3,9) * 0.4
)

print('PHQ-9 severity distribution:')
print(df['phq9_severity'].value_counts())
print(f'\nAt-risk students (PHQ-9≥10 or GAD-7≥10): {df["at_risk"].sum()} ({df["at_risk"].mean()*100:.1f}%)')
print(f'Adequate sleep (≥7h): {df["sleep_adequate"].sum()} ({df["sleep_adequate"].mean()*100:.1f}%)')
```

## 4. Depression (PHQ-9) — Severity & Distribution <a id='4'></a>

```python
fig, axes = plt.subplots(1, 3, figsize=(19, 7))
fig.patch.set_facecolor(BG)
fig.suptitle('PHQ-9 Depression Scores — 500 Students',
             fontsize=16, fontweight='bold', color='white', y=1.01)

# 1. Score distribution histogram with severity shading
ax = axes[0]
n, bins, patches = ax.hist(df['phq9_score'].dropna(), bins=28,
                             edgecolor='none', alpha=0.9, zorder=3)
shade_ranges = [(0,4,'#3d4a5c'),(5,9,'#fbbf24'),(10,14,'#fb923c'),
                (15,19,'#f87171'),(20,27,'#dc2626')]
for lo, hi, color in shade_ranges:
    ax.axvspan(lo, hi+0.99, color=color, alpha=0.10, zorder=1)
for patch, left in zip(patches, bins[:-1]):
    for lo, hi, color in shade_ranges:
        if lo <= left <= hi: patch.set_facecolor(color); break
ax.axvline(df['phq9_score'].mean(),   color='white', lw=2, ls='--',
           label=f'Mean: {df["phq9_score"].mean():.1f}')
ax.axvline(df['phq9_score'].median(), color=SLP, lw=2, ls=':',
           label=f'Median: {df["phq9_score"].median():.0f}')
for lo, _, color in shade_ranges:
    ax.axvline(lo, color=color, lw=0.7, alpha=0.5, zorder=2)
ax.set_xlabel('PHQ-9 Score'); ax.set_ylabel('Count')
ax.set_title('PHQ-9 Score Distribution\n(shaded by severity band)', fontweight='bold')
ax.legend(fontsize=9)
ax.grid(axis='y', zorder=0); ax.set_axisbelow(True)

# 2. Severity donut
ax = axes[1]
sev_order = ['None','Mild','Moderate','Mod-Severe','Severe']
sev_counts = df['phq9_severity'].value_counts().reindex(sev_order).fillna(0)
sev_cols   = [SEV_PAL[s] for s in sev_order]
wedges, texts, ats = ax.pie(
    sev_counts.values, labels=sev_order, colors=sev_cols,
    autopct='%1.1f%%', startangle=90, pctdistance=0.72,
    wedgeprops=dict(linewidth=2.5, edgecolor='#16161d'),
    textprops=dict(fontsize=9, color='#c8c8d8'))
for at in ats: at.set_color('white'); at.set_fontsize(9); at.set_fontweight('bold')
centre = plt.Circle((0,0), 0.5, fc='#16161d')
ax.add_patch(centre)
ax.text(0, 0.08, '56.6%', ha='center', fontsize=16, fontweight='bold', color=DEP)
ax.text(0, -0.18, 'Any Symptoms', ha='center', fontsize=9, color='#7c7c96')
ax.set_title('PHQ-9 Severity Split', fontweight='bold')

# 3. PHQ-9 by gender and year
ax = axes[2]
years = [1,2,3,4,5]
for gender, color in GENDER_PAL.items():
    means = [df[(df['year_of_study']==yr)&(df['gender']==gender)]['phq9_score'].mean()
             for yr in years]
    ax.plot(years, means, 'o-', color=color, lw=2.2, ms=8, label=gender, zorder=5)
ax.set_xlabel('Year of Study'); ax.set_ylabel('Mean PHQ-9 Score')
ax.set_title('PHQ-9 by Year & Gender', fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, zorder=0); ax.set_axisbelow(True)
ax.set_xticks(years)

plt.tight_layout()
plt.savefig('02_phq9.png', dpi=150, bbox_inches='tight', facecolor=BG)
plt.show()
```

## 5. Anxiety (GAD-7) & Stress Scale Analysis <a id='5'></a>

```python
fig, axes = plt.subplots(1, 3, figsize=(19, 7))
fig.patch.set_facecolor(BG)
fig.suptitle('Anxiety (GAD-7) & Perceived Stress Scale',
             fontsize=16, fontweight='bold', color='white', y=1.01)

# 1. GAD-7 + PHQ-9 KDE overlaid
ax = axes[0]
df['phq9_score'].dropna().plot.kde(ax=ax, color=DEP, lw=2.5,
    label=f'PHQ-9 μ={df["phq9_score"].mean():.1f}')
df['gad7_score'].dropna().plot.kde(ax=ax, color=ANX, lw=2.5,
    label=f'GAD-7 μ={df["gad7_score"].mean():.1f}')
df['perceived_stress_scale'].dropna().div(4).plot.kde(
    ax=ax, color=STR, lw=2.2, ls='--',
    label=f'PSS/4 μ={df["perceived_stress_scale"].mean()/4:.1f}')
ax.axvline(10, color='white', lw=1.2, ls=':', alpha=0.5,
           label='Clinical threshold (10)')
ax.set_xlabel('Score'); ax.set_ylabel('Density')
ax.set_title('Mental Health Score Distributions\n(PSS scaled /4 for comparison)',
             fontweight='bold')
ax.legend(fontsize=9); ax.grid(True, zorder=0)

# 2. GAD-7 vs PHQ-9 scatter — coloured by PSS
ax = axes[1]
sc = ax.scatter(df['phq9_score'], df['gad7_score'],
                c=df['perceived_stress_scale'], cmap=CMAP_DEP,
                s=20, alpha=0.55, zorder=3)
plt.colorbar(sc, ax=ax, label='Perceived Stress Scale', pad=0.02)
m, b, r, _, _ = stats.linregress(
    df['phq9_score'].dropna(), df['gad7_score'].dropna())
xr = np.linspace(0, 27, 80)
ax.plot(xr, m*xr+b, color='white', lw=2, zorder=6, label=f'r={r:.3f}')
ax.axvline(10, color=DEP, lw=1, ls='--', alpha=0.5)
ax.axhline(10, color=ANX, lw=1, ls='--', alpha=0.5)
ax.set_xlabel('PHQ-9 Score'); ax.set_ylabel('GAD-7 Score')
ax.set_title(f'PHQ-9 vs GAD-7  (r={r:.3f})\nColoured by Perceived Stress',
             fontweight='bold')
ax.legend(fontsize=10)
ax.text(18, 18, 'High distress\nzone', ha='center', fontsize=8.5,
        color=DEP, bbox=dict(boxstyle='round', facecolor='#16161d',
                              edgecolor=DEP, alpha=0.85))
ax.grid(True, zorder=0); ax.set_axisbelow(True)

# 3. PSS by semester type (box)
ax = axes[2]
sems = ['Fall','Spring','Summer']
sem_data = [df[df['semester_type']==s]['perceived_stress_scale'].dropna() for s in sems]
sem_cols  = [DEP, ACT, STR]
bp = ax.boxplot(sem_data, patch_artist=True,
                medianprops=dict(color='white', linewidth=2.5),
                whiskerprops=dict(color=GRID), capprops=dict(color=GRID),
                flierprops=dict(marker='o', markersize=3,
                                markerfacecolor=GRID, alpha=0.4, linestyle='none'))
for patch, c in zip(bp['boxes'], sem_cols): patch.set_facecolor(c); patch.set_alpha(0.55)
ax.set_xticklabels([f'{s}\n(n={len(d)})' for s, d in zip(sems, sem_data)], fontsize=10)
ax.set_ylabel('Perceived Stress Scale')
ax.set_title('Stress by Semester Type', fontweight='bold')
ax.grid(axis='y', zorder=0); ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig('03_anxiety_stress.png', dpi=150, bbox_inches='tight', facecolor=BG)
plt.show()
```

## 12. ML Baseline — Predicting Depression Severity <a id='12'></a>

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (classification_report, confusion_matrix,
                              roc_auc_score, f1_score)

# Binary target: at_risk (PHQ-9≥10 or GAD-7≥10)
feat_cols = ['age','year_of_study','daily_sleep_hours','daily_screen_time_hours',
             'physical_activity_days_per_week','weekly_study_hours',
             'class_attendance_percent','online_learning_satisfaction',
             'financial_stress','cgpa','gender_enc','job_enc','sem_enc']

df_ml = df_enc[feat_cols + ['at_risk']].dropna()
X, y  = df_ml[feat_cols], df_ml['at_risk']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)

scaler  = StandardScaler()
X_tr_s  = scaler.fit_transform(X_train)
X_ts_s  = scaler.transform(X_test)

models = [
    ('Logistic Regression',  LogisticRegression(random_state=42, max_iter=500), True),
    ('Random Forest',        RandomForestClassifier(n_estimators=200, random_state=42), False),
    ('Gradient Boosting',    GradientBoostingClassifier(n_estimators=200, random_state=42), False),
]

results = {}
for name, model, scaled in models:
    Xtr = X_tr_s if scaled else X_train.values
    Xts = X_ts_s if scaled else X_test.values
    model.fit(Xtr, y_train)
    preds = model.predict(Xts)
    proba = model.predict_proba(Xts)[:,1]
    auc   = roc_auc_score(y_test, proba)
    f1    = f1_score(y_test, preds)
    results[name] = {'model':model,'preds':preds,'proba':proba,'auc':auc,'f1':f1}
    print(f'\n=== {name} ===')
    print(classification_report(y_test, preds, target_names=['Not At-Risk','At-Risk']))
    print(f'  ROC-AUC: {auc:.4f} | F1: {f1:.4f}')
```

```python
from sklearn.metrics import roc_curve

fig, axes = plt.subplots(1, 3, figsize=(19, 7))
fig.patch.set_facecolor(BG)
fig.suptitle('At-Risk Prediction — ROC · Confusion · Feature Importance',
             fontsize=16, fontweight='bold', color='white', y=1.01)

model_cols = [ACA, ACT, ANX]

# 1. ROC
ax = axes[0]
ax.plot([0,1],[0,1], color='#3d4a5c', lw=1.2, ls='--')
for (name, res), color in zip(results.items(), model_cols):
    fpr, tpr, _ = roc_curve(y_test, res['proba'])
    ax.plot(fpr, tpr, color=color, lw=2.5, label=f'{name}\n(AUC={res["auc"]:.3f})')
ax.set_xlabel('False Positive Rate'); ax.set_ylabel('True Positive Rate')
ax.set_title('ROC Curves — Clinical Risk Prediction', fontweight='bold')
ax.legend(fontsize=8.5); ax.grid(True, zorder=0)

# 2. Confusion matrix (best)
ax = axes[1]
best = max(results, key=lambda k: results[k]['auc'])
cm   = confusion_matrix(y_test, results[best]['preds'])
im   = ax.imshow(cm, cmap=CMAP_DEP, aspect='auto')
ax.set_xticks([0,1]); ax.set_xticklabels(['Pred: Safe','Pred: At-Risk'], fontsize=11)
ax.set_yticks([0,1]); ax.set_yticklabels(['True: Safe','True: At-Risk'], fontsize=11)
ax.set_title(f'Confusion Matrix\n{best}', fontweight='bold')
for i in range(2):
    for j in range(2):
        ax.text(j, i, str(cm[i,j]), ha='center', va='center',
                fontsize=22, fontweight='bold', color='white')
plt.colorbar(im, ax=ax, pad=0.02)

# 3. Feature importance
ax = axes[2]
rf = results['Random Forest']['model']
imp = pd.Series(rf.feature_importances_, index=feat_cols).sort_values()
clr = [DEP if f in ['cgpa','gad7_score','phq9_score'] else
       (SLP if 'sleep' in f else
        (FIN if 'financial' in f else ACA))
       for f in imp.index]
ax.barh(range(len(imp)), imp.values, color=clr,
         height=0.65, zorder=3, edgecolor='none', alpha=0.9)
ax.set_yticks(range(len(imp)))
ax.set_yticklabels([f.replace('_',' ').title() for f in imp.index], fontsize=9)
ax.set_xlabel('Gini Importance')
ax.set_title('Feature Importance\n(Random Forest)', fontweight='bold')
for bar, v in zip(ax.patches, imp.values):
    ax.text(bar.get_width()+0.001, bar.get_y()+bar.get_height()/2,
            f'{v:.3f}', va='center', fontsize=8.5, color='white')
ax.grid(axis='x', zorder=0); ax.set_axisbelow(True)

plt.tight_layout()
plt.savefig('10_ml_results.png', dpi=150, bbox_inches='tight', facecolor=BG)
plt.show()
```
